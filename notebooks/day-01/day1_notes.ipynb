## 1. Learning Paradigms / 学习范式

### Supervised learning / 监督学习

Definition:

- Learn from labelled examples $(x,y)$.

中文：

- 给模型很多“题目 + 标准答案”，让它学会映射。

Example:

- Input: image of digit.
- Label: digit class.

### Unsupervised learning / 无监督学习

Definition:

- Find patterns without explicit labels.

Example:

- Cluster similar player behaviors.

### Reinforcement learning / 强化学习

Definition:

- Agent acts, environment responds, agent receives reward.

Example:

- Atari agent learns to maximize game score.

**Common mistake:** supervised learning learns from correct labels; RL learns from consequences of actions.


## 2. Regression and Loss Functions / 回归与损失函数

Linear regression:

$$
\hat{y} = ax + b
$$

where:

- $a$ is slope / 斜率,
- $b$ is intercept / 截距,
- $\hat{y}$ is prediction / 预测值.

Mean squared error:

$$
L(\theta)=\frac{1}{N}\sum_{i=1}^N(\hat{y}_i-y_i)^2
$$

中文解释：

- 每个点预测错多少？平方。
- 所有错误平均一下。
- loss 越小，模型越贴近数据。

Example:

- If true value is 10 and prediction is 8, error is -2 and squared error is 4.


````python
def mean_squared_error(predictions, targets):
    errors = [(p - y) ** 2 for p, y in zip(predictions, targets)]
    return sum(errors) / len(errors)


predictions = [2.0, 4.0, 6.0]
targets = [1.0, 5.0, 6.0]
print("MSE:", mean_squared_error(predictions, targets))
````

## 3. Gradient Descent / 梯度下降

Update rule:

$$
\theta_{t+1} = \theta_t - \alpha \nabla L(\theta_t)
$$

Vocabulary:

- $\theta$: parameters / 参数
- $\alpha$: learning rate / 学习率
- $\nabla L$: gradient of loss / 损失函数梯度

中文解释：

- gradient 指向 loss 增长最快的方向。
- 我们要让 loss 变小，所以往负梯度方向走。

Example:

- If the slope is too small, gradient descent adjusts it upward.

Failure modes:

- learning rate too small: slow.
- learning rate too large: overshoot or divergence.


````python
# Tiny one-parameter gradient descent example:
# minimize L(w) = (w - 3)^2
w = 0.0
learning_rate = 0.2

for step in range(8):
    gradient = 2 * (w - 3)
    w = w - learning_rate * gradient
    print(f"step={step} w={w:.4f} loss={(w - 3) ** 2:.4f}")
````

## 4. Logistic Regression / 逻辑回归

Logistic regression is for binary classification.

Score:

$$
z = w^\top x + b
$$

Sigmoid:

$$
\sigma(z)=\frac{1}{1+e^{-z}}
$$

Prediction:

$$
P(y=1\mid x)=\sigma(z)
$$

中文解释：

- linear regression 输出任意实数。
- logistic regression 把输出压到 0 到 1，当作概率。

Example:

- Spam classifier: probability that email is spam.


## 5. Automatic Differentiation and Backpropagation / 自动微分与反向传播

Why needed?

- Small models: humans can manually derive gradients.
- Large neural networks: manual gradients are impossible in practice.

Methods:

| Method | 中文 | Meaning |
| --- | --- | --- |
| manual differentiation | 手动求导 | Human writes derivative. |
| numerical differentiation | 数值求导 | Approximate derivative by small perturbation. |
| symbolic differentiation | 符号求导 | Algebraic expression manipulation. |
| automatic differentiation | 自动微分 | Program records operations and applies chain rule exactly/numerically. |

Reverse-mode AD:

- efficient when many parameters and one scalar loss,
- this is why it is central in deep learning.

Backpropagation:

- neural-network version of reverse-mode automatic differentiation.

小白例子：

- forward pass: 算答案和 loss。
- backward pass: 回头算每个参数该怎么改。


## 6. Neural Networks / 神经网络

A neural network is a layered function approximator.

Simple structure:

```text
input -> linear layer -> activation -> linear layer -> output
```

Why activation matters:

- without nonlinearity, stacked linear layers are still just one linear function.
- activation lets the network approximate more complex functions.

Example:

- For Atari, a CNN can convert image pixels into useful features.

Professional note:

- Neural networks do not magically understand the task.
- They optimize the objective we define.


## 7. RL Foundations / 强化学习基础

Basic loop:

```text
observe state/observation
    -> choose action
    -> environment changes
    -> receive reward
    -> repeat
```

中文：

- agent 像玩家。
- environment 像游戏世界。
- reward 像分数反馈。
- policy 像玩家习惯/策略。

Key challenge:

- reward can be delayed.

Example:

- In Boxing, one dodge may not score immediately but can prevent future loss.


## 8. MDP and Markov Property / MDP 与马尔可夫性质

MDP tuple:

$$
(\mathcal{S}, \mathcal{A}, P, R, \gamma)
$$

Meaning:

- $\mathcal{S}$: states / 状态集合
- $\mathcal{A}$: actions / 动作集合
- $P$: transition probability / 转移概率
- $R$: reward function / 奖励函数
- $\gamma$: discount factor / 折扣因子

Markov property:

$$
P(s_{t+1}\mid s_t,a_t,\text{history})=P(s_{t+1}\mid s_t,a_t)
$$

中文解释：

- 如果 state 真的完整，那么过去的信息已经被 state 包含了。

Example:

- Chess board state is almost enough for legal moves.
- A single Atari frame may not show velocity, so stacked frames can help.


## 9. Reward, Return, and Expectation / 奖励、回报与期望

Immediate reward:

$$
r_t
$$

Discounted return:

$$
G_t = r_t+\gamma r_{t+1}+\gamma^2r_{t+2}+\cdots
$$

中文解释：

- reward 是眼前一步。
- return 是未来总收益。
- expectation 是“平均意义上的结果”，因为环境/策略可能有随机性。

Example:

- A risky move may sometimes win big and sometimes fail; expected return averages those possibilities.


## 10. Value and Q-functions / 状态价值与动作价值

State value:

$$
V^\pi(s)=\mathbb{E}_\pi[G_t\mid s_t=s]
$$

Action value:

$$
Q^\pi(s,a)=\mathbb{E}_\pi[G_t\mid s_t=s,a_t=a]
$$

中文区别：

- $V(s)$: 这个局面本身好不好？
- $Q(s,a)$: 在这个局面做这个动作好不好？

Example:

- $V(s)$: "Standing near opponent is dangerous."
- $Q(s,\text{punch})$: "Punching now may score."


## 11. Q-learning and DQN / Q 学习与深度 Q 学习

Bellman target:

$$
y = r+\gamma\max_{a'}Q(s',a')
$$

Q-learning update:

$$
Q(s,a)\leftarrow Q(s,a)+\alpha(y-Q(s,a))
$$

DQN:

$$
Q_\theta(s,a)
$$

中文解释：

- Q-learning 用表格存 Q。
- DQN 用 neural network 估计 Q。

Example:

- Atari pixels are too many for a table, so DQN uses a CNN or neural network.


````python
def bellman_target(reward, gamma, next_q_values, done):
    if done:
        return reward
    return reward + gamma * max(next_q_values)


print("target:", bellman_target(1.0, 0.99, [0.2, 2.0, 0.5], False))
print("terminal target:", bellman_target(1.0, 0.99, [0.2, 2.0, 0.5], True))
````

## 12. On-policy, Off-policy, and Offline RL / 同策略、异策略、离线 RL

| Type | 中文 | Meaning | Example |
| --- | --- | --- | --- |
| on-policy | 同策略 | Learn from data collected by current policy. | PPO rollout data. |
| off-policy | 异策略 | Learn from data collected by another/older policy. | DQN replay buffer. |
| offline RL | 离线 RL | Learn from fixed dataset without new interaction. | logged game trajectories. |

小白例子：

- on-policy: 我边玩边用当前打法学习。
- off-policy: 我从以前的录像/旧策略经验里学。
- offline: 我只能看已有录像，不能再进游戏试。


## 13. Model-free vs Model-based RL / 无模型与有模型 RL

Model-free:

- learns policy/value directly,
- does not explicitly learn environment transition model.

Model-based:

- learns or uses a model of environment dynamics,
- can plan using predicted future.

Example:

- Model-free: DQN learns Q-values from experience.
- Model-based: agent learns "if I move right, ball moves like this" and plans.

Common mistake:

- "model" in model-based RL means environment dynamics model, not just any neural network.


## 14. Policy Gradients, Actor-Critic, and PPO / 策略梯度、Actor-Critic 与 PPO

Policy gradient:

- directly updates the policy to increase expected return.

Actor-critic:

- actor chooses actions,
- critic estimates value and gives learning signal.

PPO:

- conservative policy update,
- avoids changing policy too much at once,
- often uses clipped objective.

中文解释：

- actor 像“玩家”。
- critic 像“教练/评分员”。
- PPO 像“别一下子改打法太猛，否则崩掉”。


## 15. Exploration vs Exploitation / 探索与利用

Exploration:

- try actions you are unsure about.

Exploitation:

- use the best-known action.

Example:

- A Boxing agent may try random dodges early to learn what works.
- Later it exploits actions with high Q-values.

Common mistake:

- Too much exploration is random and inefficient.
- Too little exploration may never discover good strategies.

## 17. Common Confusions / 常见混淆

| Confusion | Correct distinction / 正确区分 |
| --- | --- |
| reward vs return | reward is immediate; return is long-term cumulative reward. |
| state vs observation | state is complete; observation is what agent receives. |
| value vs Q-value | value scores state; Q-value scores state-action pair. |
| supervised learning vs RL | supervised has labels; RL has rewards from actions. |
| model-free vs neural network | model-free can still use a neural network; it means no environment dynamics model. |
| on-policy vs off-policy | on-policy learns from current policy; off-policy can learn from old/other data. |
| gradient vs update | gradient says direction of increase; update uses negative gradient to reduce loss. |
| loss vs reward | supervised learning minimizes loss; RL often maximizes return. |
